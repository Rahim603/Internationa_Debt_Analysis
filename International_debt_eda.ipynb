EDA Step — Outlier analysis

Step 18 — Calculate IQR

```python
Q1 = country_debt_2024_clean["Value"].quantile(0.25)
Q3 = country_debt_2024_clean["Value"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

Calculating upper outlier boundary

```python
upper_bound = Q3 + (1.5 * IQR)

print("Upper outlier boundary:", upper_bound)
```

identify the countries above that boundary:

```python
outliers = country_debt_2024_clean[
    country_debt_2024_clean["Value"] > upper_bound
].sort_values("Value", ascending=False)

outliers[["Country Name", "Country Code", "Value"]]
```

2024 external debt outliers

Countries whose 2024 external debt exceeds the IQR upper-outlier boundary of $104.64 billion.

```python
import matplotlib.pyplot as plt

# Sort countries from highest to lowest debt
outliers_sorted = outliers.sort_values("Value", ascending=True)

plt.figure(figsize=(12, 8))

plt.barh(
    outliers_sorted["Country Name"],
    outliers_sorted["Value"]
)

plt.title("2024 External Debt Outliers")
plt.xlabel("External Debt (Current US$)")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

same graph with deeper analysis

```python
import matplotlib.pyplot as plt

outliers_sorted = outliers.sort_values("Value", ascending=True)

plt.figure(figsize=(12, 8))

plt.barh(
    outliers_sorted["Country Name"],
    outliers_sorted["Value"]
)

plt.axvline(
    upper_bound,
    linestyle="--",
    label=f"Outlier Threshold: ${upper_bound/1e9:.2f}B"
)

plt.title("2024 External Debt Outliers")
plt.xlabel("External Debt (Current US$)")
plt.ylabel("Country")
plt.legend()

plt.tight_layout()
plt.show()
```

Step 6 — Prepare debt_data

```python
# Prepare data for debt_data table

debt_df = df[
    [
        'Country Code',
        'Series Code',
        'Counterpart-Area Name',
        'Counterpart-Area Code',
        'Year',
        'Value'
    ]
].copy()

print("Debt data prepared!")
print("Rows:", len(debt_df))
print("Columns:", debt_df.columns.tolist())
```

Step 7 — Insert debt_data

```python
# Insert debt data into MySQL in batches

# Get country IDs
cursor.execute("SELECT country_id, country_code FROM countries")
country_map = {code: country_id for country_id, code in cursor.fetchall()}

# Get indicator IDs
cursor.execute("SELECT indicator_id, series_code FROM indicators")
indicator_map = {code: indicator_id for indicator_id, code in cursor.fetchall()}

print("Countries mapped:", len(country_map))
print("Indicators mapped:", len(indicator_map))


# Prepare insertion query
debt_insert_query = """
INSERT INTO debt_data (
    country_id,
    indicator_id,
    counterpart_area_name,
    counterpart_area_code,
    year,
    debt_value
)
VALUES (%s, %s, %s, %s, %s, %s)
"""


# Convert DataFrame rows into MySQL records
debt_data = [
    (
        country_map[row[0]],
        indicator_map[row[1]],
        row[2],
        row[3],
        int(row[4]),
        float(row[5])
    )
    for row in debt_df.itertuples(index=False, name=None)
]


# Insert in batches
batch_size = 10000

for start in range(0, len(debt_data), batch_size):
    batch = debt_data[start:start + batch_size]

    cursor.executemany(debt_insert_query, batch)
    conn.commit()

    print(
        f"Inserted {min(start + batch_size, len(debt_data)):,} "
        f"/ {len(debt_data):,} rows"
    )

print("✅ Debt data inserted successfully!")
```
