```python
#Filling numeric column with median
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
#Filling categorical column with mode
categorical_cols = df.select_dtypes(include=["object"]).columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
import pandas as pd

# Load the CSV
cleaned_path = r"C:\Users\LENOVO\Documents\Project_2_International_Debt_Analysis\IDS_FootNoteMetaData_Cleaned.csv"

df_footnote = pd.read_csv(cleaned_path)

# Separate Country and Country Code
df_footnote[["Country", "Country Code"]] = (
    df_footnote["Country Code"]
    .astype(str)
    .str.extract(r"^(.*?)\s*\(([^()]*)\)\s*$")
)

# Separate Year and Time Code
df_footnote[["Year", "Time Code"]] = (
    df_footnote["Time Code"]
    .astype(str)
    .str.extract(r"^(\d{4})\s*\(([^()]*)\)\s*$")
)

# Convert Year to integer
df_footnote["Year"] = pd.to_numeric(
    df_footnote["Year"],
    errors="coerce"
).astype("Int64")

# Arrange columns
df_footnote = df_footnote[
    [
        "Type",
        "Country",
        "Country Code",
        "Series Code",
        "Year",
        "Time Code",
        "Description"
    ]
]

# Save
df_footnote.to_csv(cleaned_path, index=False)

# Display
print("Cleaning completed successfully!")
print(df_footnote.head(10))
```
